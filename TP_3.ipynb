## 1. Buscar elemento en una colección

### Paradigma Procedural
Uso un ciclo para recorrer la lista preguntando en cada posición si encuentro el valor, comprobando esto con una variable “encontrado”.

````python
def buscar_procedural(lista, objetivo):
    encontrado = False
    for i in range(len(lista)):
        if lista[i] == objetivo:
            encontrado = True
            break
    return encontrado
````

### Paradigma Orientado a Objetos
El objeto lista ya sabe cómo buscar si contiene “algo”. El comportamiento está encapsulado dentro del objeto.

````python
mi_lista = [1, 3, 7, 9]
resultado = mi_lista.__contains__(7)

# Prueba del código
print("¿Contiene el elemento?:", resultado)
````

### Paradigma Funcional
Usando funciones que reciben otras funciones como argumentos. Más matemático u declarativo.

````python
lista = [1, 3, 7, 9]
resultado = list(filter(lambda x: x == 7, lista))
existe = len(resultado) > 0

# Prueba del código
print("¿Existe el elemento?:", existe)
````

## 2. Ordenar elementos en una colección

### Paradigma Procedural
Bubble Sort – Debo aclararle a la máquina cómo debe mover cada número/ítem.

````python
def ordenar_procedural(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                temporal = lista[j]
                lista[j] = lista[j + 1]
                lista[j + 1] = temporal
    return lista
````

### Paradigma Orientado a Objetos
Sabiendo que lista es un objeto que ya contiene los métodos incorporados. El objeto se encarga de su comportamiento (modifica la lista original).

````python
mi_lista = [5, 2, 9, 1]
mi_lista.sort()

# Prueba del código
print("Lista modificada por objeto:", mi_lista)
````

### Paradigma Funcional
No modifico la lista, sino que creo una nueva pero ordenada (Inmutabilidad).

````python
lista_original = [5, 2, 9, 1]
lista_ordenada = sorted(lista_original)

# Prueba del código
print("Original:", lista_original)
print("Nueva lista ordenada:", lista_ordenada)
````

### Paradigma Lógico
Al no existir un comando para ordenar, hay que definir lo que significa en nuestro contexto que la lista esté ordenada (Ejemplo conceptual en Prolog).

```prolog

ordenada([_]).                  % “Una lista de solo UN elemento está ordenada”
ordenada([X, Y | Cola]) :-      % “Una lista está ordenada si el primer elemento es menor o igual...”
    X =< Y,
    ordenada([Y | Cola]).

## 3. Gestionar una lista de tareas pendientes

### Paradigma Procedural
Usando diccionarios y funciones globales que modifican el estado de la lista.

````python
tareas = []

def agregar_tarea(titulo, prioridad):
    tarea = {"titulo": titulo, "prioridad": prioridad, "completada": False}
    tareas.append(tarea)

def marcar_completada(indice):
    tareas[indice]["completada"] = True

def mostrar_pendientes():
    pendientes = [t for t in tareas if not t["completada"]]
    pendientes.sort(key=lambda t: t["prioridad"])
    for t in pendientes:
        print(f"[{t['prioridad']}] {t['titulo']}")

# Prueba rápida
agregar_tarea("Estudiar", 1)
agregar_tarea("Ir al super", 3)
mostrar_pendientes()
````

### Paradigma Orientado a Objetos
Creo una clase Tarea que conoce los datos que contiene y cómo alterar sus estados, y un Gestor que administra la colección.

````python
class Tarea:
    def __init__(self, titulo, prioridad):
        self.titulo = titulo
        self.prioridad = prioridad
        self.completada = False

    def completar(self):
        self.completada = True

class GestorTareas:
    def __init__(self):
        self.tareas = []

    def agregar(self, tarea):
        self.tareas.append(tarea)

    def listar_pendientes(self):
        pendientes = [t for t in self.tareas if not t.completada]
        return sorted(pendientes, key=lambda t: t.prioridad)

# Prueba rápida
gestor = GestorTareas()
gestor.agregar(Tarea("Comprar cuadernillos", 2))
for t in gestor.listar_pendientes():
    print(f"[{t.prioridad}] {t.titulo}")
````

### Paradigma Funcional
Funciones puras que reciben colecciones y devuelven nuevas colecciones sin alterar los estados previos.

````python
def agregar_tarea(lista, titulo, prioridad):
    nueva_tarea = (titulo, prioridad, False)
    return lista + [nueva_tarea]

def completar_tarea(lista, titulo):
    return [ (t[0], t[1], True) if t[0] == titulo else t for t in lista ]

def obtener_pendientes(lista):
    pendientes = filter(lambda t: t[2] == False, lista)
    return sorted(pendientes, key=lambda t: t[1])

# Prueba rápida
lista_tareas = []
lista_tareas = agregar_tarea(lista_tareas, "Ir al almacén", 3)
print(obtener_pendientes(lista_tareas))
````
