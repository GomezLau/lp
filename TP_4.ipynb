# Definición del Lenguaje y Gramáticas

### Características Generales del Lenguaje

* **Tipo de datos:** ENTERO
* **Identificadores:** * Implícitos y de **MÁXIMO 4 CARACTERES**.
  * Deben comenzar con una letra y estar compuestos por letras, dígitos y guiones bajos (`_`).
  * **Restricciones:** NO pueden terminar con `_` ni contener dos guiones bajos seguidos (`__`).
* **Constantes:** Secuencias de dígitos (enteros).
* **Sentencias:**
  * **Asignación:** `ID ::= Expressao;`
  * **Entrada/Salida:** `ler(lista de IDs);` o `escrever(lista de Expressoes);`
  * *Nota:* Cada sentencia termina obligatoriamente con punto y coma `;`.
* **Cuerpo del programa:** Delimitado por las palabras clave `comeco -> final`.
* **Palabras reservadas:** `comeco`, `final`, `ler`, `escrever`.

## 1. Gramática Independiente del Contexto (GIC)

* **En (No Terminales):** `{programa, lista_sentencias, sentencia, asignacion, entrada_salida, lista_ids, lista_expressao, expressao, factor, constante, identificador, caracter, caracter_interno, letra, digito}`
* **Et (Terminales):** `{comeco, final, ler, (, ), escrever, ::=, +, -, _, ;, ,}`
* **S (Axioma):** `programa`
* **P (Producciones):**

```text
programa -> comeco lista_sentencias final
lista_sentencias -> sentencia lista_sentencias | sentencia
sentencia -> asignacion | entrada_salida
asignacion -> identificador ::= expressao ;
entrada_salida -> ler(lista_ids) ; | escrever(lista_expressao) ;
lista_ids -> identificador | identificador, lista_ids
lista_expressao -> expressao | expressao, lista_expressao
expressao -> factor | expressao + factor | expressao - factor
factor -> identificador | constante | (expressao)
constante -> digito | digito constante
identificador -> letra | letra caracter | letra caracter_interno caracter | letra caracter_interno caracter caracter | letra caracter caracter_interno caracter
caracter_interno -> _ | caracter
caracter -> letra | digito
letra -> A|...|Z | a|...|z
digito -> 0|...|9

## 2. Forma Backus-Naur (BNF)

```
<programa> ::= comeco <lista_sentencias> final
<lista_sentencias> ::= <sentencia> <lista_sentencias> | <sentencia>
<sentencia> ::= <asignacion> | <entrada_salida>
<asignacion> ::= <identificador> ::= <expressao> ;
<entrada_salida> ::= ler( <lista_ids> ) ; | escrever( <lista_expressao> ) ;

<lista_ids> ::= <identificador> | <identificador> , <lista_ids>
<lista_expressao> ::= <expressao> | <expressao> , <lista_expressao>

<expressao> ::= <factor> | <expressao> + <factor> | <expressao> - <factor>
<factor> ::= <identificador> | <constante> | ( <expressao> )
<constante> ::= <digito> | <digito> <constante>

<identificador> ::= <letra> | <letra> <caracter> | <letra> <caracter_interno> <caracter> | <letra> (<caracter_interno> <caracter> | <caracter> <caracter_interno> ) <caracter>
<caracter> ::= <letra> | <digito>
<caracter_interno> ::= _ | <caracter>
<letra> ::= a|...|z | A|...|Z
<digito> ::= 0|...|9

```




## 3. Forma de Backus-Naur Extendida (EBNF)

```
<programa> ::= comeco <lista_sentencias> final
<lista_sentencias> ::= <sentencia> { <sentencia> }
<sentencia> ::= <asignacion> | <entrada_salida>
<asignacion> ::= <identificador> ::= <expressao> ;
<entrada_salida> ::= ler( <lista_ids> ) ; | escrever( <lista_expressao> ) ;

<lista_ids> ::= <identificador> { , <identificador> }
<lista_expressao> ::= <expressao> { , <expressao> }

<expressao> ::= <factor> { (+|-) <factor> }*
<factor> ::= <identificador> | <constante> | ( <expressao> )
<constante> ::= <digito> { <digito> }

<identificador> ::= <letra> [ <caracter> ] [ <caracter> ] [ <caracter> ] | <letra> _ <caracter> [<caracter>] | <letra> <caracter> _ <caracter>
<caracter> ::= <letra> | <digito>
<letra> ::= a|...|z | A|...|Z
<digito> ::= 0|...|9

```



## 4. Forma Backus-Naur Aumentada (ABNF)

```
programa : comeco lista_sentencias final
lista_sentencias : sentencia { lista_sentencias }
sentencia : asignacion | entrada_salida
asignacion : identificador ::= expressao ;
entrada_salida : ler( lista_ids ) ; | escrever( lista_expressao ) ;

lista_ids : identificador { , identificador }
lista_expressao : expressao { , expressao }

expressao : factor { (uno de + -) factor }
factor : uno de identificador constante (expressao)
constante : digito {constante}

identificador : letra caracter _op caracter _op caracter _op
                letra _ caracter caracter _op
                letra caracter _ caracter
caracter : uno de letra digito
letra : uno de a...z A...Z
digito : uno de 0...9
```
