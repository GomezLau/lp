Defina un LP para un campo de aplicaciones particular, con la característica de que este LP esté formado por un grupo de comandos que son LRs. Describa su sintaxis en BNF, su semántica en lenguaje natural y presente varios ejemplos para su uso

```python
DOMO-LANG es un lenguaje de programacion especializado en la domotica, con el proposito principal de permitir el control de distintos electrodomesticos.
Lenguaje propuesto: DOMO-LANG
Tipos de datos: entero.
Identificadores: Deben comenzar con una letra y pueden contener letras, dígitos o guiones.
Palabras reservadas: inicio, fin, ENCENDER, APAGAR, CONFIGURAR son palabras reservadas.
Hay dos tipos de sentencias:

Asignacion
ID := Valor;

Entrada/Salida
ENCENDER( lista de IDs);
APAGAR( lista de IDs);
CONFIGURAR( lista de IDs, valor );

BNF
<programa> ::= inicio <lista_sentencias> fin
<lista_sentencias> ::= <sentencia> | <sentencia> <lista_sentencias>
<sentencia> ::= <asignacion> | <entrada_salida>
<asignacion> ::= <identificador> := <valor> ;
<entrada_salida> ::= ENCENDER(<lista_ids>) ; | APAGAR(<lista_ids>) ; | CONFIGURAR(<lista_ids>, <valor>)  ; 
<lista_ids> ::= <identificador> | <identificador> , <lista_ids>

<valor> ::= <identificador> | <constante>
<constante> ::= <digito> | <digito> <constante>

<identificador> ::= <letra> | <letra> <caracter> | <letra> <caracter_interno> <caracter> | <letra> (<caracter_interno> <caracter> | <caracter> <caracter_interno>) <caracter>
<caracter> ::= <letra> | <digito>
<caracter_interno> ::= _ | <caracter>
<letra> ::= a|...|z | A|...|Z
<digito> ::= 0|...|9

inicio y fin delimitan el comienzo y la finalizacion del programa
ENCENDER(<lista_ids>) Cambia el estado de los dispositivos en la lista a "Encendido"
APAGAR(<lista_ids>) Cambia el estado de los dispositivos en la lista a "Apagado"
CONFIGURAR(<lista_ids>,<valor>) cambia la potencia o intensidad de los dispositivos en la lista al valor especificado.

# Ejemplo de Rutina de Inicio de Dia
# El sistema ya reconoce luz_cocina y pava_electrica como dispositivos de la casa
# hervir es un identificador que apunta al valor 100
inicio 
    ENCENDER(luz_cocina, pava_electrica);
    CONFIGURAR(luz_cocina, 50);
    CONFIGURAR(pava_electrica, hervir);
fin

# Ejemplo de Rutina de Fin de Dia
# El sistema ya reconoce luz_cocina, luz_living y alarma como dispositivos de la casa
inicio 
    APAGAR(luz_cocina, luz_living);
    ENCENDER(alarma);
fin

# Ejemplo de Rutina de Llegada a Casa
# El sistema ya reconoce luz_cocina, luz_living, alarma y aire_acondicionado como dispositivos de la casa
inicio 
    APAGAR(alarma);
    ENCENDER(luz_cocina, luz_living);
    CONFIGURAR(aire_acondicionado, 24);
fin

```

En Python se habla de tipos de datos mutables e inmutables. ¿Para que se utiliza este tipo de datos? De al menos un ejemplo donde se puede utilizar ¿Cómo? ¿Sucede lo mismo para Ruby? Explique la forma en que un objeto mutable puede convertirse en inmutable

```python
Los tipos de datos mutables se utilizan cuando el contenido del dato debe ser modificable sin cambiar su direccion en memoria.
Por ejemplo una lista de compras tiene una coleccion de datos que van a sufrir cambios.
Los tipos inmutables se usan para garantizar que los datos sean compartidos y seguros.
Como no van a cambiar el lenguaje puede optimizar el uso de memoria, como por ejemplo claves de diccionarios.
Un tipo inmutable podria ser definir los dias de la semana o las coordenadas de un lugar fijo.

En Python la inmutabilidad de un objeto es determinada por el tipo de dato (un string siempre es inmutable)
Pero en Ruby casi todos los objetos son mutables por defecto.

En Python un objeto mutable por ejemplo una lista puede convertirse en tupla mediante una funcion y convertirse a inmutable
En Ruby existe un metodo del lenguaje que permite bloquear un objeto mutable para que nada lo pueda modificar.
".freeze"
La clase "Data" de Ruby permite que una vez inicializada no se puedan modificar los atributos de ese objeto.
```
